```python
import pandas as pd
import numpy as np
```

### Load the dataset


```python
df = pd.read_csv("Luxury_Housing_Bangalore.csv")
```

```python

print(df.columns)
```

```python
df.info()
```

```python
#checking for null values
df.isnull().sum()
```

```python
#cheching for duplicate values
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
(df.select_dtypes(include="number") < 0).any()
```

```python


# Replace invalid unit sizes with NaN
df.loc[df["Unit_Size_Sqft"] <= 0, "Unit_Size_Sqft"] = np.nan

# Fill missing values using median of the same configuration
df["Unit_Size_Sqft"] = (
    df.groupby("Configuration")["Unit_Size_Sqft"]
      .transform(lambda x: x.fillna(x.median()))
)
```

```python
#Clean text columns
df["Micro_Market"] = df["Micro_Market"].str.title()
```

```python
df["Configuration"] = (
    df["Configuration"]
    .str.replace(" ", "", regex=False)
    .str.upper()
)
```

```python
df['Configuration'].value_counts()
```

```python
df["NRI_Buyer"] = df["NRI_Buyer"].str.title()
```

```python
df["Ticket_Price_Cr"] = (
    df["Ticket_Price_Cr"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("Cr", "", regex=False)
    .str.strip()
)

df["Ticket_Price_Cr"] = pd.to_numeric(
    df["Ticket_Price_Cr"], 
    errors="coerce"
)
```

```python
df["Ticket_Price_Cr"].isnull().sum()
```

```python
(df.select_dtypes(include="number") < 0).any()
```

```python
df.loc[df["Ticket_Price_Cr"] <= 0, "Ticket_Price_Cr"] = np.nan
```

```python
#filling null values in Ticket_Price_Cr column with mean value of the column
df["Ticket_Price_Cr"].median()
```

```python
numeric_columns = [
    "Ticket_Price_Cr",
    "Unit_Size_Sqft",
    "Amenity_Score"
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

df["Buyer_Comments"] = df["Buyer_Comments"].fillna("No comment")
```

```python
#Convert Purchase_Quarter to Date
df["Purchase_Quarter"].dtype
```

```python
df["Purchase_Quarter"] = pd.to_datetime(
    df["Purchase_Quarter"],
    errors="coerce"
)
```

```python
df["Purchase_Year"] = df["Purchase_Quarter"].dt.year
```

```python
df["Quarter_Number"] = df["Purchase_Quarter"].dt.quarter
```

```python
df["Quarter_Label"] = (
    df["Purchase_Year"].astype(str)
    + "-Q"
    + df["Quarter_Number"].astype(str)
)
```

```python
df.shape
```

### Feature Engineering

```python
##Price_per_Sqft
df["Price_per_Sqft"] = (
    (df["Ticket_Price_Cr"] * 10000000) / df["Unit_Size_Sqft"]
).round(2)
```

```python
df[[
    "Ticket_Price_Cr",
    "Unit_Size_Sqft",
    "Price_per_Sqft"
]].head(10)
```

```python
df["Price_per_Sqft"].describe()
```

```python
(df.select_dtypes(include="number") < 0).any()
```

```python
#Nbooking Flag
df["Booking_Flag"] = df["Transaction_Type"].map(
    {"Primary": 1, "Secondary": 0}
)
```

```python
df["Booking_Flag"].value_counts()
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("Null values:", df.isnull().sum().sum())
print("Duplicate values:", df.duplicated().sum())
print("Negative values:", (df.select_dtypes(include="number") < 0).any().any())
```

```python
df.to_csv(
    "Luxury_Housing_Bangalore_Cleaned.csv",
    index=False
)
```
