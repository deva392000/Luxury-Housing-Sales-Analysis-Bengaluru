```python
import pandas as pd
import pyodbc

```

### Load cleaned CSV & SQL Server connection

```python
# Load cleaned CSV
df = pd.read_csv("Luxury_Housing_Bangalore_Cleaned.csv")

# SQL Server connection
conn = pyodbc.connect(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=DESKTOP-55JNORK\SQLEXPRESS;"
    "DATABASE=LuxuryHousing;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

cursor = conn.cursor()

print("Connected successfully!")
```

```python
df.shape
```

```python
insert_query = """

INSERT INTO Luxury_Housing (
    Property_ID,
    Micro_Market,
    Project_Name,
    Developer_Name,
    Unit_Size_Sqft,
    Configuration,
    Ticket_Price_Cr,
    Transaction_Type,
    Buyer_Type,
    Purchase_Quarter,
    Connectivity_Score,
    Amenity_Score,
    Possession_Status,
    Sales_Channel,
    NRI_Buyer,
    Locality_Infra_Score,
    Avg_Traffic_Time_Min,
    Buyer_Comments,
    Purchase_Year,
    Quarter_Number,
    Quarter_Label,
    Price_per_Sqft,
    Booking_Flag )
VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
"""


# Convert DataFrame rows into tuples
data = list(df.itertuples(index=False, name=None))

# Fast insert
cursor.fast_executemany = True
cursor.executemany(insert_query, data)

# Save changes
conn.commit()

print("Successfully inserted!")
```

### Read the SQL table

```python
query = """
SELECT *
FROM Luxury_Housing;
"""

df_sql = pd.read_sql(query, conn)

print("Rows and Columns:", df_sql.shape)
```

```python
# Create final CSV
df_sql.to_csv(
    "Luxury_Housing_Bangalore_FINAL.csv",
    index=False
)

print("Final CSV created successfully!")
```
